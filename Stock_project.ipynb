```python
import requests
import json
import os

API_KEY = "3ESSIL2OI3XS3MEQ"
SYMBOL = "IBM"

folder = "alpha_vantage_data"
os.makedirs(folder, exist_ok=True)

# -----------------------------
# 1. Company Overview
# -----------------------------
url_overview = f"https://www.alphavantage.co/query?function=OVERVIEW&symbol={SYMBOL}&apikey={API_KEY}"
print("API Endpoint (Overview):", url_overview)

response_overview = requests.get(url_overview)
overview_data = response_overview.json()

print("\nCompany Overview JSON:\n", overview_data)

with open(f"{folder}/company_overview.json", "w") as f:
    json.dump(overview_data, f, indent=4)

# -----------------------------
# 2. Historical Stock Data
# -----------------------------
url_history = f"https://www.alphavantage.co/query?function=TIME_SERIES_DAILY&symbol={SYMBOL}&apikey={API_KEY}"
print("\nAPI Endpoint (Historical):", url_history)

response_history = requests.get(url_history)
history_data = response_history.json()

print("\nHistorical Data JSON:\n", history_data)

with open(f"{folder}/historical_data.json", "w") as f:
    json.dump(history_data, f, indent=4)

# -----------------------------
# 3. Financial / Fundamental Data
# -----------------------------
url_financials = f"https://www.alphavantage.co/query?function=INCOME_STATEMENT&symbol={SYMBOL}&apikey={API_KEY}"
print("\nAPI Endpoint (Financials):", url_financials)

response_financials = requests.get(url_financials)
financial_data = response_financials.json()

print("\nFinancial Data JSON:\n", financial_data)

with open(f"{folder}/financial_data.json", "w") as f:
    json.dump(financial_data, f, indent=4)

print("\nAll API responses saved successfully.")
```
